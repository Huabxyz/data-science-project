# Historical Tech Stock Price Analysis (2013–2018)

**Research question:** How did the historical closing prices and day-to-day variability of Apple, Amazon, Facebook, Google, and Microsoft compare between February 2013 and February 2018?

Based on the original `mc648026_Activity6.01.ipynb` activity, with normalized prices, volatility comparison, and limitations added.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd()
SYMBOLS = ["GOOGL", "FB", "AAPL", "AMZN", "MSFT"]
closing = pd.DataFrame({s: pd.read_csv(ROOT / "data" / f"{s}_data.csv", parse_dates=["date"]).set_index("date")["close"] for s in SYMBOLS}).sort_index().dropna()
closing.head()
```

## Figure 1: Indexed closing prices
Each stock starts at 100 so relative price changes are easier to compare. **This is not a split-adjusted or dividend-adjusted return comparison.**

```python
indexed = closing / closing.iloc[0] * 100
ax = indexed.plot(figsize=(11,6), title="Indexed closing prices (first day = 100)")
ax.set_ylabel("Indexed closing price")
plt.tight_layout(); plt.show()
```

## Figure 2: Daily closing-price return volatility
Volatility here is the standard deviation of one-day percentage changes in closing prices.

```python
daily_returns = closing.pct_change().dropna()
vol = daily_returns.std() * 100
ax = vol.sort_values().plot.barh(figsize=(9,5), title="Daily close-to-close volatility")
ax.set_xlabel("Standard deviation of daily percentage change (%)")
plt.tight_layout(); plt.show()
```

```python
pd.DataFrame({"Price change (%)":(closing.iloc[-1]/closing.iloc[0]-1)*100, "Daily volatility (%)":vol}).round(2).sort_values("Price change (%)",ascending=False)
```

## Findings and limitations
- In these supplied **raw closing-price series**, FB has the greatest first-to-last percentage increase (~531%), followed by AMZN (~441%).
- FB also has the greatest daily percentage-change standard deviation (~2.01%).
- **Limitations:** The files do not establish adjusted prices or dividend treatment. Stock splits/corporate actions could distort comparisons. The sample ends in February 2018 and does not represent recent market conditions. Dataset license/provenance needs confirmation before public redistribution.
